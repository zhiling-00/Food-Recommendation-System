```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv(r"C:\Users\Zhi Ling.DELL-ZLPG\Downloads\recipes.csv")
```

```python
print(df.head())       # View the first few rows
print(df.info())       # Get a concise summary of the DataFrame
```

```python
df.drop(columns=['prep_time', 'cook_time', 'servings', 'yield', 'url', 'img_src', 'timing', 'cuisine_path', 'Unnamed: 0'], inplace=True)
```

```python
df.rename(columns={'directions': 'instructions', 'ingredients': 'ingredients details'}, inplace=True)
```

```python
print(df.info())  
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.loc[15, 'total_time'] = "3 hrs 15 mins"
```

```python
def convert_to_minutes(time):
    if pd.isna(time):
        return np.nan
    parts = time.split()
    total_minutes = 0
    for i in range(0, len(parts), 2):
        value = int(parts[i])
        unit = parts[i+1]
        if 'hrs' in unit:
            total_minutes += value * 60
        elif 'mins' in unit:
            total_minutes += value
    return total_minutes

df['total_time_minutes'] = df['total_time'].apply(convert_to_minutes)

# Calculate mean
mean_minutes = df['total_time_minutes'].mean()

# Fill missing values with the mean
df['total_time_minutes'].fillna(mean_minutes, inplace=True)

def convert_to_hours_minutes(minutes):
    hours = minutes // 60
    mins = minutes % 60
    if hours == 0:
        return f"{int(mins)} mins"
    else:
        return f"{int(hours)} hrs {int(mins)} mins"

df['total_time_filled'] = df['total_time_minutes'].apply(convert_to_hours_minutes)
```

```python
df.drop(columns=['total_time_minutes', 'total_time'], inplace=True)
```

```python
print("Cleaned Dataset:")
print(df.head())
print(df.info())   
```

```python
df.to_excel("cleaned_recipe.xlsx", index=False, engine='openpyxl')
```
